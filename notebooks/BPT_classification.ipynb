```python
nsa_dict = {}
for i in range(len(nsa_portsmouth)):
    nsa_dict[nsa_portsmouth['NSAID'][i]] = i
```

```python
def BPT(Ha, Hb, N2, O3):
    '''
    1: SF
    2: comp
    3: AGN
    '''

    O3Hb = np.log10(O3/Hb)
    N2Ha = np.log10(N2/Ha)

    if N2Ha > 0.4:
        return 3

    if O3Hb < 0.61/(N2Ha - 0.05) + 1.3 and N2Ha < 0.01:
        return 1

    elif (0.61/(N2Ha - 0.05) + 1.3 < O3Hb and 0.61/(N2Ha - 0.47) + 1.19 > O3Hb) or (N2Ha > 0 and 0.61/(N2Ha - 0.47) + 1.19 > O3Hb):
        return 2

    else:
        return 3

def dust_corr(Ha, Hb, N2, O3, corr_law='CCM89'):
    
    rc= pn.RedCorr(law=corr_law)

    H_ratio = Ha/Hb

    rc.setCorr(H_ratio / 2.86, 6562, 4861)

    # in order Ha, Hb, NII, OIII
    wavelengths = np.array([6562, 4861, 6583, 5006])
    corr = rc.getCorrHb(wavelengths)

    Ha_corr = Ha * corr[0]
    Hb_corr = Hb * corr[1]
    N2_corr = N2 * corr[2]
    O3_corr = O3 * corr[3]

    return Ha_corr, Hb_corr, N2_corr, O3_corr
```

```python
# add BPT flag to master table
# add flux ratios

master_table['BPT_class'] = 0.
master_table['N2Ha'] = -999 * np.ones(len(master_table))
master_table['O3Hb'] = -999 * np.ones(len(master_table))

for i in range(len(master_table)):
# for i in range(10):


    # check if gal is in nsa
    if master_table['nsa_nsaid'][i] >= 0:

        # find gal in portsmouth table
        nsaid = master_table[i]['nsa_nsaid']
        nsa_idx = nsa_dict[nsaid]

        # check that all necessary fluxes are > 0

        Ha = nsa_portsmouth['Flux_Ha_6562'][nsa_idx]
        Hb = nsa_portsmouth['Flux_Hb_4861'][nsa_idx]
        N2 = nsa_portsmouth['Flux_NII_6583'][nsa_idx]
        O3 = nsa_portsmouth['Flux_OIII_5006'][nsa_idx]

        if Ha > 0 and Hb > 0 and N2 > 0 and O3 > 0:


            # dust correct 
            Ha_corr, Hb_corr, N2_corr, O3_corr = dust_corr(Ha, Hb, N2, O3)

            # print(Ha_corr, Hb_corr, N2_corr, O3_corr)

            
            O3Hb = np.log10(O3_corr/Hb_corr)
            N2Ha = np.log10(N2_corr/Ha_corr)

            # print(O3Hb)
            # print(N2Ha)
            
            master_table['BPT_class'][i] = BPT(Ha_corr,Hb_corr,N2_corr,O3_corr)
            master_table['N2Ha'][i] = N2Ha
            master_table['O3Hb'][i] = O3Hb
            
```

```python
plt.scatter(master_table['O3Hb'][master_table['BPT_class'] == 1], 
            master_table['N2Ha'][master_table['BPT_class'] == 1], 
            color='b', marker='.', s=0.5, label='sta)
plt.scatter(master_table['O3Hb'][master_table['BPT_class'] == 2], 
            master_table['N2Ha'][master_table['BPT_class'] == 2], 
            color='g', marker='.', s=0.5)
plt.scatter(master_table['O3Hb'][master_table['BPT_class'] == 3], 
            master_table['N2Ha'][master_table['BPT_class'] == 3], 
            color='r', marker='.', s=0.5)
plt.scatter(master_table['O3Hb'][master_table['BPT_class'] == 0], 
            master_table['N2Ha'][master_table['BPT_class'] == 0], 
            color='k', marker='.', s=0.5)

plt.xlim(-2,2)
plt.ylim(-2,2)

plt.xlabel(r'$\log([\rm OIII]/\rm H\beta)$')
plt.ylabel(r'$\log([\rm NII]/\rm H\alpha)$')
```
