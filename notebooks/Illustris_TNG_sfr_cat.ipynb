```python
import numpy as np
import illustris_python as il
import h5py as h5
from astropy.table import Table
from tqdm import tqdm
```

```python
basePath = '/home/tnguser/sims.TNG/TNG100-1/output/'
fields = ['SubhaloMass','SubhaloSFRinRad']
subhalos = il.groupcat.loadSubhalos(basePath, 99, fields=fields)
```

```python
sfr_fn = '/home/tnguser/sims.TNG/TNG100-1/postprocessing/released/star_formation_rates.hdf5'
sfr = h5.File(sfr_fn, 'r')
```

```python
sfr_99 = sfr['Snapshot_99']
```

```python
list(sfr_99.keys())
```

```python
SubfindIDs = np.array(sfr_99['SubfindID'])
SFR_all_10Myr = np.array(sfr_99['SFR_MsunPerYrs_in_all_10Myrs'])
SFR_2R50_10Myr = np.array(sfr_99['SFR_MsunPerYrs_in_InRad_10Myrs'])
```

```python
flag = np.ones(len(SubfindIDs))*-999
Vmax = np.ones(len(SubfindIDs))*-999
SubhaloMass = np.ones(len(SubfindIDs))*-999
SubhaloMassInRadStar = np.ones(len(SubfindIDs))*-999
SubhaloHalfmassRadStar = np.ones(len(SubfindIDs))*-999
SubhaloStellarPhotometricsMassInRad = np.ones(len(SubfindIDs))*-999
SubhaloStellarPhotometricsRad = np.ones(len(SubfindIDs))*-999
```

```python
for i in tqdm(range(len(SubfindIDs))):
# for i in tqdm(range(100)):


    subhalo = il.groupcat.loadSingle(basePath,99,subhaloID=SubfindIDs[i])
    
    flag[i] = subhalo['SubhaloFlag']
    Vmax[i] = subhalo['SubhaloVmax']
    SubhaloMass[i] = subhalo['SubhaloMass']
    SubhaloMassInRadStar[i] = subhalo['SubhaloMassInRadType'][4]
    SubhaloHalfmassRadStar[i] = subhalo['SubhaloHalfmassRadType'][4]
    SubhaloStellarPhotometricsMassInRad[i] = subhalo['SubhaloStellarPhotometricsMassInRad']
    SubhaloStellarPhotometricsRad[i] = subhalo['SubhaloStellarPhotometricsRad']
```

```python
from astropy.io import fits
```

```python
IDs = fits.Column(name='SubfindID', format='32A', array=SubfindIds)
hdu = fits.BinTableHDU.from_columns([IDs])
hdu.writeto('/home/tnguser/IllustrisTNG100-1_sfr_subhalos.fits', overwrite=True)
```

```python
t = Table.read('/home/tnguser/IllustrisTNG100-1_sfr_subhalos.fits', format='fits')
```

```python
t['SubhaloFlag'] = flag
t['SubhaloVmax'] = Vmax
t['SubhaloMass'] = SubhaloMass
t['SubhaloMassInRadStar'] = SubhaloMassInRadStar
t['SubhaloHalfmassRadStar'] = SubhaloHalfmassRadStar
t['SubhaloStellarPhotometricsMassInRad'] = SubhaloStellarPhotometricsMassInRad
t['SubjaloStellarPhotometricsRad'] = SubhaloStellarPhotometricsMassInRad
t['SFR_10Myr'] = SFR_all_10Myr
t['SFR_InRadStar_10Myr'] = SFR_2R50_10Myr
```

```python
t.write('/home/tnguser/IllustrisTNG100-1_sfr_subhalos.fits', overwrite=True)
```
