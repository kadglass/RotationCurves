```python
import numpy as np
from astropy.table import Table
import matplotlib.pyplot as plt
from astropy.io import fits
import matplotlib
```

```python
matplotlib.rcParams['figure.dpi'] = 100
matplotlib.rcParams['savefig.dpi'] = 300
```

```python
DATA_DIR = '/Users/nityaravi/Documents/Research/RotationCurves/data/manga/'
master_table_fn = DATA_DIR + 'output_files/DR17/CURRENT_MASTER_TABLE/Elliptical_sphdisk_refitspirals_v2.fits'
dynpop_fn = DATA_DIR + 'dynpop/SDSSDR17_MaNGA_JAM.fits'
plot_dir = DATA_DIR + 'Ellipticals_Images/results_plots/refit_spirals/'
```

```python
master_table = Table.read(master_table_fn, format='fits')
master_table[:5]
```

```python
dynpop_1 = Table.read(dynpop_fn, format='fits', hdu=1) #JAM sph w NFW
dynpop_1[:5]
```

```python
dynpop_dict = {}
for i in range(len(dynpop_1)):
    dynpop_dict[dynpop_1['plateifu'][i]] = i
```

```python
dynpop = Table.read(dynpop_fn, format='fits', hdu=5) #JAM sph w NFW
dynpop[:5]
```

```python
ellipticals = master_table[np.logical_and(master_table['Mvir'] > 0, master_table['sphd_M_star']>0)]
```

```python
spirals = master_table[master_table['v_max'] > 0]
spirals = spirals[spirals['M90_disk'] > 0]
spirals = spirals[spirals['alpha'] <= 99]
spirals = spirals[spirals['v_max_err']/spirals['v_max'] <= 2]
v90 = spirals['v_max'] * spirals['R90_kpc'] / (spirals['r_turn']**spirals['alpha'] + spirals['R90_kpc']**spirals['alpha'])**(1/spirals['alpha'])
all_mass_spirals = spirals[np.logical_and(v90 > 10, v90 < 1000)]
spirals = all_mass_spirals[all_mass_spirals['M90_disk']>9]
```

```python
for i in range(len(ellipticals)):

    gal_ID = ellipticals['plateifu'][i]
    d_idx = dynpop_dict[gal_ID]
    
    plt.plot(ellipticals['Mvir'][i], dynpop['log_Mt_Re'][d_idx], 
                linestyle='none', 
                marker='.', color='g', alpha=0.4, label='within 2D $R_{50}$')
    # plt.plot(ellipticals['Mvir'][i], dynpop['log_Mt_rhalf'][d_idx], 
    #             linestyle='none', 
    #             marker='.', color='orange', alpha=0.4, label='within 3D $R_{50}$')


plt.plot([9,14],[9,14], color='k')
handles, labels = plt.gca().get_legend_handles_labels()
by_label = dict(zip(labels, handles))
plt.legend(by_label.values(), by_label.keys())

plt.xlabel(r'$M{\rm vir}\ [\log(M_\odot)]$')
plt.ylabel(r'DynPop $M_{\rm tot}\ [\log(M_\odot)]$') 
plt.savefig(plot_dir + 'Mvir_dynpopMtot_ellipticals_scatter.png')
```

```python
H_0 = 100      # Hubble's Constant in units of h km/s/Mpc
c = 299792.458 # Speed of light in units of km/s
G = 4.30091E-3 # Gravitation constant in units of (km/s)^2 pc/Msun
MANGA_SPAXEL_SIZE = 0.5*(1/60)*(1/60)*(np.pi/180)  # spaxel size (0.5") in radians


dist_to_galaxy_Mpc = c*spirals['z']/H_0
dist_to_galaxy_kpc = dist_to_galaxy_Mpc*1000
pix_scale_factor = dist_to_galaxy_kpc*np.tan(MANGA_SPAXEL_SIZE)
```

```python
for i in range(len(spirals)):

    gal_ID = spirals['plateifu'][i]
    d_idx = dynpop_dict[gal_ID]

    R50_2D = dynpop_1['Re_arcsec_MGE'][d_idx]
    R50_3D = dynpop['rhalf_arcsec'][d_idx]
    
    v50_2D = spirals['v_max'][i] * R50_2D / (spirals['r_turn'][i]**spirals['alpha'][i] 
                                             + R50_2D**spirals['alpha'][i])**(1/spirals['alpha'][i])

    
    v50_3D = spirals['v_max'][i] * R50_3D / (spirals['r_turn'][i]**spirals['alpha'][i] 
                                             + R50_3D**spirals['alpha'][i])**(1/spirals['alpha'][i])

    M50_2D = np.log10(v50_2D**2 * R50_2D / G)

    M50_3D = np.log10(v50_3D**2 * R50_3D / G)
    
    plt.plot(M50_2D, dynpop['log_Mt_Re'][d_idx], 
                linestyle='none', 
                marker='.', color='g', alpha=0.4, label='within 2D $R_{50}$')
    plt.plot(M50_3D, dynpop['log_Mt_rhalf'][d_idx], 
                linestyle='none', 
                marker='.', color='orange', alpha=0.4, label='within 3D $R_{50}$')


plt.plot([9,14],[9,14], color='k')
handles, labels = plt.gca().get_legend_handles_labels()
by_label = dict(zip(labels, handles))
plt.legend(by_label.values(), by_label.keys())

plt.xlabel(r'$M{\rm vir}\ [\log(M_\odot)]$')
plt.ylabel(r'DynPop $M_{\rm tot}\ [\log(M_\odot)]$') 
```
