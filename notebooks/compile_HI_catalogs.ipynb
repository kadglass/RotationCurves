# Create master HI catalog
extract HI info from all HI-MaNGA versions, MaNGA-ALFALFA DR15 catalog, and ALFALFA-SDSS cross match

#### imports, data paths

```python
from astropy.table import Table
from astropy.io import fits
import numpy as np
```

```python
DATA_DIRECTORY = '/Users/nityaravi/Documents/Research/RotationCurves/data/'
master_table_fn = DATA_DIRECTORY + 'manga/output_files/DR17/CURRENT_MASTER_TABLE/Elliptical_sphdisk_refitspirals_BPT_v8.fits'
manga_alfalfa_dr15_fn = DATA_DIRECTORY + '/manga/manga_alfalfa-dr15.fits'
hi_manga_dr3_fn = DATA_DIRECTORY + 'manga/mangaHIall.fits'
hi_manga_dr4_fn = DATA_DIRECTORY + 'manga/himanga_dr4.fits'
alfalfa_sdss_fn = DATA_DIRECTORY + 'durbala2020-table2.21-Sep-2020.fits'
alfalfa_sdss_crossmatch_fn = DATA_DIRECTORY + 'durbala2020-table1.21-Sep-2020.fits'
dl_fn = DATA_DIRECTORY + '/manga/manga-morphology-dl-DR17.fits'
```

```python
# load master_table
master_table = Table.read(master_table_fn)
```

```python
plateifu = fits.Column(name='plateifu', format='32A', array=master_table['plateifu'])
```

```python
hdu = fits.BinTableHDU.from_columns([plateifu])
```

```python
hdu.writeto(DATA_DIRECTORY + 'HI_manga_alfalfa_mastertable.fits', overwrite=True)
```

```python
hi_master = Table.read(DATA_DIRECTORY + 'HI_manga_alfalfa_mastertable.fits')
```

```python
hi_master[:5]
```

```python
hi_master_dict = {}
for i in range(len(hi_master)):
    hi_master_dict[hi_master['plateifu'][i]] = i
```

### manga_alfalfa-dr15

```python
manga_alfalfa = Table.read(manga_alfalfa_dr15_fn)
manga_alfalfa[:5]
```

```python
hi_master['MHI_manga_alfalfa_dr15'] = np.ones(len(hi_master)) * -999
hi_master['MHI_err_manga_alfalfa_dr15'] = np.ones(len(hi_master)) * -999

for i in range(len(manga_alfalfa)):

    gal_id = manga_alfalfa['PLATEIFU'][i].split(' ')[0]
    try:
        idx = hi_master_dict[gal_id]
    except:
        print(gal_id, ' not found in dict')
        continue

    hi_master['MHI_manga_alfalfa_dr15'][idx] = manga_alfalfa['LOGMHI'][i]
```

### hi-manga dr3

```python
hi_manga_dr3 = Table.read(hi_manga_dr3_fn)
hi_manga_dr3[-5:]
```

```python
hi_master['MHI_himanga_dr3'] =np.ones(len(hi_master)) * -999
hi_master['MHI_err_himanga_dr3'] = np.ones(len(hi_master)) * -999
hi_master['MHI_himanga_dr3_alfalfa'] = np.ones(len(hi_master)) * -999
hi_master['MHI_err_himanga_dr3_alfalfa'] = np.ones(len(hi_master)) * -999

for i in range(len(hi_manga_dr3)):

    gal_id = hi_manga_dr3['PLATEIFU'][i].split(' ')[0]
    try:
        idx = hi_master_dict[gal_id]
    except:
        print(gal_id, ' not found in dict')
        continue

    if hi_manga_dr3['SESSION'][i] == 'ALFALFA':
    
        hi_master['MHI_himanga_dr3_alfalfa'][idx] = hi_manga_dr3['LOGMHI'][i]
        if hi_manga_dr3['LOGMHI'][i] > 0:
            hi_master['MHI_err_himanga_dr3_alfalfa'][idx] = hi_manga_dr3['LOGMHI'][i] \
                                                            + np.log10(hi_manga_dr3['EFHI'][i]) - np.log10(hi_manga_dr3['FHI'][i])
    else:
        hi_master['MHI_himanga_dr3'][idx] = hi_manga_dr3['LOGMHI'][i]

        if hi_manga_dr3['LOGMHI'][i] > 0:
        
            hi_master['MHI_err_himanga_dr3'][idx] = hi_manga_dr3['LOGMHI'][i] \
                                                            + np.log10(hi_manga_dr3['EFHI'][i]) - np.log10(hi_manga_dr3['FHI'][i])
   
        
```

### hi-manga dr4

```python
hi_manga_dr4 = Table.read(hi_manga_dr4_fn)
hi_manga_dr4[:5]
```

```python
hi_master['MHI_himanga_dr4'] =np.ones(len(hi_master)) * -999
hi_master['MHI_err_himanga_dr4'] = np.ones(len(hi_master)) * -999
hi_master['MHI_himanga_dr4_alfalfa'] = np.ones(len(hi_master)) * -999
hi_master['MHI_err_himanga_dr4_alfalfa'] = np.ones(len(hi_master)) * -999

for i in range(len(hi_manga_dr4)):

    gal_id = hi_manga_dr4['PLATEIFU'][i].split(' ')[0]
    try:
        idx = hi_master_dict[gal_id]
    except:
        print(gal_id, ' not found in dict')
        continue

    if hi_manga_dr4['SESSION'][i] == 'ALFALFA':
    
        hi_master['MHI_himanga_dr4_alfalfa'][idx] = hi_manga_dr4['LOGMHI'][i]
        if hi_manga_dr4['LOGMHI'][i] > 0:
            hi_master['MHI_err_himanga_dr4_alfalfa'][idx] = hi_manga_dr4['LOGMHI'][i] \
                                                            + np.log10(hi_manga_dr4['EFHI'][i]) - np.log10(hi_manga_dr4['FHI'][i])
    else:
        hi_master['MHI_himanga_dr4'][idx] = hi_manga_dr3['LOGMHI'][i]

        if hi_manga_dr3['LOGMHI'][i] > 0:
        
            hi_master['MHI_err_himanga_dr4'][idx] = hi_manga_dr4['LOGMHI'][i] \
                                                            + np.log10(hi_manga_dr4['EFHI'][i]) - np.log10(hi_manga_dr4['FHI'][i])
   
        
```

### selection
loop through table and select best mass measurement in this order
hi_manga_dr4 > hi_manga_dr_3 > hi_manga_dr4_alfalfa > hi_manga_dr3_alfalfa > manga_alfalfa

```python
hi_master['logHI'] = np.ones(len(hi_master))*-999
hi_master['logHI_err'] = np.ones(len(hi_master))*-999
hi_master['HI_catalog'] = ["                                       " for x in range(len(master_table))]

for i in range(len(master_table)):

    MHI = -999
    MHI_err = -999
    cat = ''

    if hi_master['MHI_himanga_dr4'][i] > 0:
        MHI = hi_master['MHI_himanga_dr4'][i]
        MHI_err = hi_master['MHI_err_himanga_dr4'][i]
        cat = 'himanga_dr4'
        

    elif hi_master['MHI_himanga_dr3'][i] > 0:
        MHI = hi_master['MHI_himanga_dr3'][i]
        MHI_err = hi_master['MHI_err_himanga_dr3'][i]
        cat = 'himanga_dr3'
        

    elif hi_master['MHI_himanga_dr4_alfalfa'][i] > 0:
        MHI = hi_master['MHI_himanga_dr4_alfalfa'][i]
        MHI_err = hi_master['MHI_err_himanga_dr4_alfalfa'][i]
        cat = 'himanga_dr4_alfalfa'

    elif hi_master['MHI_himanga_dr3_alfalfa'][i] > 0:
        MHI = hi_master['MHI_himanga_dr3_alfalfa'][i]
        MHI_err = hi_master['MHI_err_himanga_dr3_alfalfa'][i]
        cat = 'himanga_dr3_alfalfa'
        

    elif hi_master['MHI_manga_alfalfa_dr15'][i] > 0:
        MHI = hi_master['MHI_manga_alfalfa_dr15'][i]
        cat = 'manga_alfalfa_dr15'
        
    hi_master['logHI'][i] = MHI
    hi_master['logHI_err'][i] = MHI_err
    hi_master['HI_catalog'][i]  = cat
```

```python
hi_master.write(DATA_DIRECTORY + 'HI_manga_alfalfa_mastertable.fits', overwrite=True, format='fits')
```
