```python
master_table['e_Mstar_R90'] = 0.
```

```python
for i in range(len(master_table)):
    if master_table['sphd_M_star'][i] > 0:

        R90 = master_table['R90_kpc'][i]
        rho_c = master_table['sphd_rho_c'][i]
        a = master_table['sphd_R_scale'][i]
        Sigd = master_table['sphd_Sigma_d'][i]
        Rd = master_table['sphd_R_d'][i]
        
        # master_table['e_Mstar_R90'][i] = np.log10(exponential_sphere(R90, rho_c, a) +
        #                                          exponential_disk(R90, Sigd, Rd)
        #     )
        master_table['e_Mstar_R90'][i] = np.log10(exponential_sphere_disk(R90, rho_c, a, Sigd, Rd))
```
