#### remove some columns

```python
master_table.remove_columns(['Mstar_HI_R90', 'Mstar_HI_3p5R90', 'Mstar_Mtot_R90','MstarHI_Mtot_R90','nsa_Mstar_Mtot',
                             'Mvis_R90', 'Mvis_Mtot_R90', 'sum_M_star_0p1_R90', 'count_M_star_0p1_R90', 
                             'Mvir2', 'Mvir2_err','star_sigma2','star_sigma2_err'])
```

#### add metallicity and age from p3d table

```python
master_table['Age_LW_Re_fit'] = -99.
master_table['Age_MW_Re_fit'] = -99.
master_table['ZH_LW_Re_fit'] = -99.
master_table['ZH_MW_Re_fit'] = -99.

for i in range(len(master_table)):

    try:

        i_p = p3d_dict[master_table[i]['plateifu']]

        master_table['Age_LW_Re_fit'][i] = p3d[i_p]['Age_LW_Re_fit']
        master_table['Age_MW_Re_fit'][i] = p3d[i_p]['Age_MW_Re_fit']
        master_table['ZH_LW_Re_fit'][i] = p3d[i_p]['ZH_LW_Re_fit']
        master_table['ZH_MW_Re_fit'][i] = p3d[i_p]['ZH_MW_Re_fit']
    
    except:
        continue
```

#### add spiral stellar mass 
within R90 in units of log(M_sun/h) ['M90_disk_h'] </p>
total stellar mass in units of log(M_sun/h) ['sphd_M_star_h']

```python
master_table['M90_disk_h'] = np.ones(len(master_table))*-999
master_table['M90_disk_err_h'] = np.ones(len(master_table))*-999
master_table['sphd_M_star_err_h'] = np.ones(len(master_table))*-999
master_table['sphd_M_star_h'] = np.ones(len(master_table))*-999

for i in range(len(master_table)):
    if master_table['M90_disk'][i] > 0:
        master_table['M90_disk_h'][i] = master_table['M90_disk'][i] + np.log10(0.73)
        master_table['M90_disk_err_h'][i] = master_table['M90_disk_err'][i] + np.log10(0.73)

        params = {'sphd_rho_c' : np.log10(master_table['rho_bulge'][i]),
                  'sphd_rho_c_err' : np.log10(master_table['rho_bulge_err'][i]),
                  'sphd_R_scale' : master_table['R_bulge'][i],
                  'sphd_R_scale_err' : master_table['R_bulge_err'][i],
                  'sphd_Sigma_d' : np.log10(master_table['Sigma_disk'][i]),
                  'sphd_Sigma_d_err' : np.log10(master_table['Sigma_disk_err'][i]),
                  'sphd_R_d' : master_table['R_disk'][i],
                  'sphd_R_d_err': master_table['R_disk_err'][i]
            
        }
        
    master_table['sphd_M_star_h'][i], master_table['sphd_M_star_err_h'][i] = np.log10(calc_tot_stellar_mass(params, 'sphere_disk')) + np.log10(0.73)
```

### add HI mass
in log(M_sun/h) units ['logHI_h']

```python
master_table['logHI_h'] = np.ones(len(master_table))*-999
for i in range(len(master_table)):
    if master_table['logHI'][i] > 0:
        master_table['logHI_h'][i] = master_table['logHI'][i] + np.log10(0.7)
```

#### calculate vr90 err for spirals

```python
master_table['V_R90'] = np.ones(len(master_table)) * -999
master_table['V_R90_err'] = np.ones(len(master_table)) * -999

for i in range(len(master_table)):

    if master_table['M_R90'][i] > 0:

        params = [master_table['v_max'][i], master_table['v_max_err'][i], master_table['r_turn'][i], master_table['r_turn_err'][i],
                  master_table['alpha'][i], master_table['alpha_err'][i]]
        master_table['V_R90'][i], master_table['V_R90_err'][i] = rot_fit_BB_err(master_table['R90_kpc'][i], params)
```

#### calculate stellar mass within r90 and error for ellipticals

```python
master_table['e_M_star_R90'] = np.ones(len(master_table)) * -999
master_table['e_M_star_R90_err'] = np.ones(len(master_table)) * -999
master_table['e_M_star_R90_h'] = np.ones(len(master_table)) * -999
master_table['e_M_star_R90_err_h'] = np.ones(len(master_table)) * -999

for i in range(len(master_table)):

    if master_table['sphd_M_star'][i] > 0:

        master_table['e_M_star_R90'][i], master_table['e_M_star_R90_err'][i] = np.log10(exponential_sphere_disk_err(master_table['R90_kpc'][i],
                                                                                                           master_table['sphd_rho_c'][i],
                                                                                                           master_table['sphd_rho_c_err'][i],
                                                                                                           master_table['sphd_R_scale'][i],
                                                                                                           master_table['sphd_R_scale_err'][i],
                                                                                                           master_table['sphd_Sigma_d'][i],
                                                                                                           master_table['sphd_Sigma_d_err'][i],
                                                                                                           master_table['sphd_R_d'][i],
                                                                                                           master_table['sphd_R_d_err'][i]))
        master_table['e_M_star_R90_h'][i] = master_table['e_M_star_R90'][i] + np.log10(0.73)
        master_table['e_M_star_R90_err_h'][i] = master_table['e_M_star_R90_err'][i] + np.log10(0.73)
```

```python
master_table.remove_columns(['e_Mstar_R90'])
```

#### calculate effective velocity at r90 and error for ellipticals

```python
master_table['v_eff'] = np.ones(len(master_table)) * -999
master_table['v_eff_err'] = np.ones(len(master_table)) * -999

for i in range(len(master_table)):

    if master_table['Mvir'][i] > 0:
        master_table['v_eff'][i], master_table['v_eff_err'][i] = calculate_effective_velocity(master_table['R90_kpc'][i],
                                                                                              master_table['Mvir'][i],
                                                                                              master_table['Mvir_err'][i])
```
