```python
# bins = np.linspace(-0.015, 0.010, 20)
bins = np.quantile(gasz['grad_Z'], np.linspace(0,1,20))
indices = np.digitize(gasz['grad_Z'], bins)

s_med = []
ctrs = []

for i in range(len(bins)):

    s_med.append(ma.median(gasz['stellar_grad_Z'][indices==i]))
    ctrs.append(ma.mean(gasz['grad_Z'][indices==i]))
```

## mstar/mtot and mvis/mtot using central stellar metallicity

```python
s_mstarmtot = spirals['sphd_M_star_h'] - spirals['M_R90']
e_mstarmtot = ellipticals['sphd_M_star_h'] - ellipticals['Mvir']
```

```python
plt.scatter(spirals['stellar_Z_0'], s_mstarmtot, alpha=0.1, marker='.')
plt.scatter(ellipticals['stellar_Z_0'], e_mstarmtot, alpha=0.1, marker='.')

plt.xlim(-1,0.4)
plt.ylim(-2,0.5)
```

```python
s_e_Z0 = ma.hstack((spirals['stellar_Z_0'], ellipticals['stellar_Z_0']))
mstarmtot = ma.hstack((s_mstarmtot, e_mstarmtot))
```

```python
bins = np.quantile(s_e_Z0[(s_e_Z0 > -1) & (s_e_Z0 < 0.4)], np.linspace(0,1,20))
# bins= np.linspace(-0.1,0.1,10)
indices = np.digitize(s_e_Z0, bins=bins)

mstarmtot_med = []
ctr = []

for i in range(len(bins)):

    mstarmtot_med.append(ma.median(mstarmtot[indices==i]))
    ctr.append(ma.mean(s_e_Z0[indices==i]))
```

```python
plt.scatter(s_e_Z0, mstarmtot, alpha=0.1, marker='.')

plt.scatter(ctr, mstarmtot_med, marker='x', color='navy')

plt.xlim(-1,0.5)
plt.ylim(-1.5,0.5)

plt.xlabel('central stellar metallicity')
plt.ylabel(r'$M_*/M_{\rm tot}$')
```

```python
HI_spirals = spirals[spirals['logHI_R90'] > 0]
```

```python
HI_spirals['param_H2'] = np.zeros(len(HI_spirals))
HI_spirals['param_H2_h'] = np.zeros(len(HI_spirals))
HI_spirals['param_He_h'] = np.zeros(len(HI_spirals))
HI_spirals['param_He90_h'] = np.zeros(len(HI_spirals))
HI_spirals['Mbar'] = np.zeros(len(HI_spirals))


for i in range(len(HI_spirals)):
    
    if HI_spirals['CMD_class'][i] == 1:
        HI_spirals['param_H2'][i] = -0.4*HI_spirals['rabsmag'][i] +1.12
    elif HI_spirals['CMD_class'][i] == 2 or HI_spirals['CMD_class'][i] == 3:
        HI_spirals['param_H2'][i] = -0.27*HI_spirals['rabsmag'][i] + 3.62

    HI_spirals['param_H2_h'][i] = HI_spirals['param_H2'][i] + np.log10(0.7)
    HI_spirals['param_He_h'][i] = np.log10(0.7) -np.log10(3) + np.log10(10**HI_spirals['logHI'][i] + 10**HI_spirals['param_H2'][i])
    HI_spirals['param_He90_h'][i] = np.log10(0.7) -np.log10(3) + np.log10(10**HI_spirals['logHI_R90'][i] + 10**HI_spirals['param_H2'][i])



for i in range(len(HI_spirals)):
    HI_spirals['Mbar'][i] = np.log10(10**(HI_spirals['M90_disk'][i] + np.log10(0.73)) + 10**(HI_spirals['logHI_R90'][i]+np.log10(0.7)) + 10**HI_spirals['param_H2_h'][i] + 10**HI_spirals['param_He90_h'][i])
```

```python
HI_mbarmtot = HI_spirals['Mbar'] - HI_spirals['M_R90']
```

```python
s_e_Z0 = ma.hstack((HI_spirals['stellar_Z_0'], ellipticals['stellar_Z_0']))
mbarmtot = ma.hstack((HI_mbarmtot, e_mstarmtot))
```

```python
bins = np.quantile(s_e_Z0[(s_e_Z0 > -1) & (s_e_Z0 < 0.4)], np.linspace(0,1,20))
# bins= np.linspace(-0.1,0.1,10)
indices = np.digitize(s_e_Z0, bins=bins)

mbarmtot_med = []
ctr = []

for i in range(len(bins)):

    mbarmtot_med.append(ma.median(mbarmtot[indices==i]))
    ctr.append(ma.mean(s_e_Z0[indices==i]))
```

```python
plt.scatter(s_e_Z0, mbarmtot, alpha=0.1, marker='.')

plt.scatter(ctr, mbarmtot_med, marker='x', color='navy')

plt.xlim(-1,0.5)
plt.ylim(-1.5,0.5)

plt.xlabel('central stellar metallicity')
plt.ylabel(r'$M_{\rm bar}/M_{\rm tot}$')
```
