```python
manga_sample_dict = {1: 'r',
                     2: 'g',
                     3: 'b',
                     0: 'k'
}
```

```python
prim = 0
sec = 0
ce = 0 
other = 0

for i in range(len(ellipticals_cut)):

    targbits = ellipticals_cut['mngtarg1'][i]

    # primary sample
    if (targbits & 1 << 1) | (targbits & 1 << 4) | (targbits & 1 << 7) | (targbits & 1 << 10):
        
        plt.scatter(ellipticals_cut['sum_M_star_0p1_R90'][i], ellipticals_cut['e_Mstar_R90'][i], edgecolor=manga_sample_dict[1],
                    color='none', alpha=0.4, marker='.')
        prim +=1

    elif (targbits & 1 << 2) | (targbits & 1 << 5) | (targbits & 1 << 8) | (targbits & 1 << 11):
        
        plt.scatter(ellipticals_cut['sum_M_star_0p1_R90'][i], ellipticals_cut['e_Mstar_R90'][i], edgecolor=manga_sample_dict[2], 
                    color='none', alpha=0.4, marker='.')

        sec += 1
        
    elif (targbits & 1 << 3) | (targbits & 1 << 6) | (targbits & 1 << 9) | (targbits & 1 << 12):
        
        plt.scatter(ellipticals_cut['sum_M_star_0p1_R90'][i], ellipticals_cut['e_Mstar_R90'][i], edgecolor=manga_sample_dict[3], 
                    color='none', alpha=0.4, marker='.')
        ce += 1

    else:

        plt.scatter(ellipticals_cut['sum_M_star_0p1_R90'][i], ellipticals_cut['e_Mstar_R90'][i], edgecolor=manga_sample_dict[0], 
                    color='none', alpha=0.4, marker='.')
        other +=1



plt.plot([0,13],[0,13], color='k', linestyle='--')
plt.xlabel(r'$M_{*, \rm spaxels}(0.1R_{90})\ [\log(M_\odot)]$')
plt.ylabel(r'$M_{*,\rm model}(R_{90})\ [\log(M_\odot)]$')
plt.xlim(7,13)
plt.ylim(7,13)

legend_elements = [Line2D([0], [0], marker='.', color='w', label='Primary Sample',
                          markeredgecolor='r', markersize=5),
                   Line2D([0], [0], marker='.', color='w', label='Secondary Sample',
                          markeredgecolor='g', markersize=5),
                   Line2D([0], [0], marker='.', color='w', label='Color-Enhanced Sample',
                          markeredgecolor='b', markersize=5)]

plt.legend(handles=legend_elements)
```
