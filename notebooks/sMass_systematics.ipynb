### model vs spax hist
sum mass: 'sum_M_star', 'sum_M_star_err' </br>
model mass: 'sphd_M_star', 'sphd_M_star_err' </br>
model mass w/in R90: 'e_Mstar_R90'

```python
# ellipticals where all cols have a value
ellipticals_cut = ellipticals[np.logical_and(ellipticals['e_Mstar_R90'] > 0, ellipticals['sum_M_star'] > 0)]
```

```python
plt.hist(np.log10(10**ellipticals_cut['sphd_M_star'] - 10**ellipticals_cut['sum_M_star']),
                  bins=np.linspace(6,14,25)
                 )
plt.xlabel(r'$\log(M_{*,\rm model} - M_{*, \rm spaxels})$')
```

```python
plt.hist(ellipticals_cut['sphd_M_star'] - ellipticals_cut['sum_M_star'], 
         bins=np.linspace(-0.5,1,25))
plt.xlabel(r'$\log(M_{*,\rm model}) - \log(M_{*, \rm spaxels})$')
```

```python
plt.hist((10**ellipticals_cut['sphd_M_star'] - 10**ellipticals_cut['sum_M_star'])/10**ellipticals_cut['sum_M_star'],
                  bins=np.linspace(-0.25,1,25)
                 )
plt.xlabel(r'$(M_{*,\rm model} - M_{*, \rm spaxels})/M_{*, \rm spaxels}$')
```

```python
plt.scatter(ellipticals_cut['sum_M_star'], ellipticals_cut['sphd_M_star'], marker='o', color='r', alpha=0.1)
plt.plot([9,18],[9,18], color='k', linestyle='--')
plt.xlabel(r'$\log(M_{*,\rm spaxels}/M_\odot)$')
plt.ylabel(r'$\log(M_{*,\rm model}/M_\odot)$')
```
