```python
#test gal
gal_ID = '8319-1902'

H_0 = 100      # Hubble's Constant in units of h km/s/Mpc
c = 299792.458 # Speed of light in units of km/s
G = 4.30091E-6 # Gravitation constant in units of (km/s)^2 kpc/Msun
MANGA_SPAXEL_SIZE = 0.5*(1/60)*(1/60)*(np.pi/180)  # spaxel size (0.5") in radians
```

### extract pipe3d map

```python
maps = extract_Pipe3d_data(PIPE3D_FOLDER, gal_ID, ['sMass'])
maps_dap = extract_data(VEL_FOLDER, gal_ID, ['flux'])
idx = master_table_dict[gal_ID]
z = master_table['nsa_z'][idx]
phi = master_table['nsa_elpetro_phi'][idx]
ba = master_table['nsa_elpetro_ba'][idx]

sMass_mask = np.isnan(maps['sMass_density'])
msMass_density = ma.array(maps['sMass_density'], mask = sMass_mask)
sMass_err_mask = np.isnan(maps['sMass_density_err'])
msMass_density_err = ma.array(maps['sMass_density_err'], mask=sMass_err_mask)

mflux = ma.array(maps_dap['mflux'], mask=sMass_mask)
optical_center = np.unravel_index(ma.argmax(mflux), mflux.shape)
r90_arcsec = master_table['nsa_elpetro_th90'][idx]
```

```python
plt.imshow(msMass_density)
plt.colorbar(label=r'stellar mass density [$\log(M_\odot)$]')
plt.gca().invert_yaxis()
plt.xlabel('spaxel')
plt.ylabel('spaxel')
```

### mask spaxels outside of ellipse with semi-major axis R

```python
dist_to_galaxy_Mpc = c*z/H_0
dist_to_galaxy_kpc = dist_to_galaxy_Mpc*1000
pix_scale_factor = dist_to_galaxy_kpc*np.tan(MANGA_SPAXEL_SIZE) 
```

```python
array_length = msMass_density.shape[0]  # y-coordinate distance
array_width = msMass_density.shape[1]  # x-coordinate distance

X_RANGE = np.arange(0, array_width, 1)
Y_RANGE = np.arange(0, array_length, 1)
X_COORD, Y_COORD = np.meshgrid( X_RANGE, Y_RANGE)
```

```python
phi_elip = (90 - phi)*np.pi/180.

x_diff = X_COORD - optical_center[1]
y_diff = Y_COORD - optical_center[0]

ellipse = (x_diff*np.cos(phi_elip) - y_diff*np.sin(phi_elip))**2 \
          + (x_diff*np.sin(phi_elip) + y_diff*np.cos(phi_elip))**2 \
          / ba**2
```

```python
R=10
ellipse_mask = ellipse > R**2
```

```python
plt.imshow(ma.array(msMass_density, mask=np.logical_or(sMass_mask, ellipse_mask)))
plt.gca().invert_yaxis()
plt.xlabel('spaxel')
plt.ylabel('spaxel')
plt.colorbar(label=r'stellar mass density [$\log(M_\odot)$]')
```

```python
center_msMass = ma.array(msMass_density, mask=np.logical_or(sMass_mask, ellipse_mask))
cent_mass = np.log10(np.sum(10**center_msMass))
```
