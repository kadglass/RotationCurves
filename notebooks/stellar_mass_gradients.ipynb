```python
def stell_mass(Rb, pb, Rd, pd, r):

    x = r / Rb
    print(x)
    
    F = 1 - np.exp(-x)*(1 + x + 0.5*x**2)
    print(F)
    
    Mb = 8 * np.pi * Rb**3 * pb * F
    print(Mb)
    
    Md = 2*np.pi*pd*Rd*(Rd - np.exp(-r/Rd)*(r +Rd))
    print(Md)
    
    return np.log10(Mb + Md)
```
