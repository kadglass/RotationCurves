## Fig 4a

```python
hdr4a = fits.Header()
hdr4a['PAPER'] = 'Ravi26'
hdr4a['FIGURE'] = '4a'

empty_primary4a = fits.PrimaryHDU(header=hdr4a)
```

```python
# read in info from text files
fn_4ae = data_folder + '/vmax/ellipticals_HI/ellipticals/data.txt'
f = open(fn_4ae, mode='r')
text = f.read()
f.close()
```

```python
idx = text.find('[')
idx2 = text.find(']')
ctrs = np.array(text[idx+1:idx2].split(',',-1)).astype(float)
text=text.split(']',1)[1]

idx = text.find('[')
idx2 = text.find(']')
avg_txt = text[idx+1:idx2]
avg_txt = avg_txt.replace(' ', ',')
avg = np.array(avg_txt.split(',',-1)).astype(float)
text=text.split(']',1)[1]

idx = text.find('[')
idx2 = text.find(']')
sig_txt = text[idx+1:idx2]
sig_txt = sig_txt.replace(' ', ',')
sig = np.array(sig_txt.split(',',-1)).astype(float)
# text=text.split(']',1)[1]
```

```python
e_data_hdu = fits.BinTableHDU(data=Table([ctrs, avg, sig], names=['X', 'Y', 'YERR']), name='ELLIPTICALS')
```

```python
# read in info from text files
fn_4as = data_folder + '/vmax/ellipticals_HI/spirals/data.txt'
f = open(fn_4as, mode='r')
text = f.read()
f.close()
```

```python
idx = text.find('[')
idx2 = text.find(']')
ctrs = np.array(text[idx+1:idx2].split(',',-1)).astype(float)
text=text.split(']',1)[1]

idx = text.find('[')
idx2 = text.find(']')
avg_txt = text[idx+1:idx2]
avg_txt = avg_txt.replace(' ', ',')
avg = np.array(avg_txt.split(',',-1)).astype(float)
text=text.split(']',1)[1]

idx = text.find('[')
idx2 = text.find(']')
sig_txt = text[idx+1:idx2]
sig_txt = sig_txt.replace(' ', ',')
sig = np.array(sig_txt.split(',',-1)).astype(float)
# text=text.split(']',1)[1]
```
