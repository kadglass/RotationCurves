```python
base_path = '/Users/nityaravi/Documents/Research/RotationCurves/data/IlustrisTNG/TNG100-1/'
fields = ['SubhaloFlag',  'SubhaloPos', 'SubhaloStellarPhotometrics', 'SubhaloVel' 
         ]
data = il.groupcat.loadSubhalos(base_path,99,fields=fields)
```

```python
good_pos = data['SubhaloPos'][data['SubhaloFlag']]
good_stel_phot = data['SubhaloStellarPhotometrics'][data['SubhaloFlag']]
good_vel = data['SubhaloVel'][data['SubhaloFlag']]
```

```python
mag_cut = good_stel_phot[:,5] < -20
```

```python
vol_good_pos = good_pos[mag_cut]
vol_good_stel_phot = good_stel_phot[mag_cut]
vol_good_vel = good_vel[mag_cut]
```

```python
col1 = fits.Column(name='SubhaloPos', format='3D', array=vol_good_pos)
col2 = fits.Column(name='SubhaloStellarPhotometrics', format='8D', array=vol_good_stel_phot)
col3 = fits.Column(name='SubhaloVel', format='3D', array=vol_good_vel)
```

```python
coldefs = fits.ColDefs([col1, col2, col3])
hdu = fits.BinTableHDU.from_columns(coldefs)
```

```python
t = Table.read(hdu)
```
